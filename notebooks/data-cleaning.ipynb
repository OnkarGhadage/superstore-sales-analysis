# Data Cleaning

## Documenting the Assessment

1. Column names should not contain the space. (Remove or replace spaces from the column names).
2. Incorrect Data types:
    1. `Order Date` and `Ship Date` columns have str dtype (Change it to datetime)
    2. `Postal Code` have int64 dtype (Change it to str)
    3. `Ship Mode`, `Segment`, `Region`, `Category`, `State` & `Sub-Category` have dtype str (Change it to Categorical)
3. Order ID `US-2014-150119` has one duplicate record. 

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw-data/superstore.csv', encoding='latin1')
```

```python
df.head()
```

# Cleaning

### Lets fix nulls

```python
# Check
df.isnull().sum()
```

### Fix Column names

```python
# Column names and clean column names dic
rename_dict = {
    'Row ID': 'row-id',
    'Order ID': 'order-id',
    'Order Date': 'order-date',
    'Ship Date': 'ship-date',
    'Ship Mode': 'ship-mode',
    'Customer ID': 'customer-id',
    'Customer Name': 'customer-name',
    'Segment': 'segment',
    'Country': 'country',
    'City': 'city',
    'State': 'state',
    'Postal Code': 'postal-code',
    'Region': 'region',
    'Product ID': 'product-id',
    'Category': 'category',
    'Sub-Category': 'sub-category',
    'Product Name': 'product-name',
    'Sales': 'sales',
    'Quantity': 'quantity',
    'Discount': 'discount',
    'Profit': 'profit'
}
```

```python
df.rename(columns=rename_dict, inplace=True)
```

```python
# Check
df.head()
```

### Fix column data types

#### Fixing `order-date` & `ship-date`

```python
df.info()
```

```python
df['order-date'] = pd.to_datetime(df['order-date'], format='%m/%d/%Y')
df['ship-date'] = pd.to_datetime(df['ship-date'], format='%m/%d/%Y')
```

```python
#Check
df.info()
```

#### Fixing `postal-code`

```python
# comvert into str
df['postal-code'] = df['postal-code'].astype('str')
```

```python
df['postal-code'].sample(5)
```

```python
# For postal code that are of length 4 we have to make it 5 by filling leading 0
df['postal-code'] = df['postal-code'].str.zfill(5)
```

```python
df['postal-code'].sample(5)
```

#### Fixing `Ship Mode`, `Segment`, `Region`, `Category`, `State` & `Sub-Category`.

```python
df['ship-mode'] = df['ship-mode'].astype('category')
df['segment'] = df['segment'].astype('category')
df['state'] = df['state'].astype('category')
df['region'] = df['region'].astype('category')
df['category'] = df['category'].astype('category')
df['sub-category'] = df['sub-category'].astype('category')
```

```python
# Check
df.info()
```
