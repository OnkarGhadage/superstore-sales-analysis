# Data Cleaning

## Documenting the Assessment

1. Column names should not contain the space. (Remove or replace spaces from the column names).
2. Incorrect Data types:
    1. `Order Date` and `Ship Date` columns have str dtype (Change it to datetime)
    2. `Postal Code` have float64 dtype (Change it to str)
    3. `Ship Mode`, `Segment`, `Region`, `Category` & `Sub-Category` have dtype str (Change it to Categorical)
3. `Postal Code` have 11 missing values (Tyr to fill it with correct values)

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw-data/train.csv')
```

```python
df.head()
```

# Cleaning

### Fix null values

```python
# Check the recored of missing postal code
df[df['Postal Code'].isnull()]
```

```python
# check any other customer is from Burlington city to get postal code  
df[df['City']=='Burlington']
```

```python
# So the Burlington from state Vermont has missing postal code
# There are in total 6 postal codes for Burlington Vermont
# 05401, 05402, 05404, 05405, 05406 & 05408
# Lets replace missing values wiht primary one (05401)
df.fillna({'Postal Code':5401}, inplace=True)
```

```python
# Check
df.isnull().sum()
```

### Fix Column names

```python
# Column names and clean column names dic
rename_dict = {
    'Row ID': 'row-id',
    'Order ID': 'order-id',
    'Order Date': 'order-date',
    'Ship Date': 'ship-date',
    'Ship Mode': 'ship-mode',
    'Customer ID': 'customer-id',
    'Customer Name': 'customer-name',
    'Segment': 'segment',
    'Country': 'country',
    'City': 'city',
    'State': 'state',
    'Postal Code': 'postal-code',
    'Region': 'region',
    'Product ID': 'product-id',
    'Category': 'category',
    'Sub-Category': 'sub-category',
    'Product Name': 'product-name',
    'Sales': 'sales'
}
```

```python
df.rename(columns=rename_dict, inplace=True)
```

```python
# Check
df.head()
```

### Fix column data types

#### Fixing `order-date` & `ship-date`

```python
df.info()
```

```python
df['order-date'] = pd.to_datetime(df['order-date'], format='%d/%m/%Y')
df['ship-date'] = pd.to_datetime(df['ship-date'], format='%d/%m/%Y')
```

```python
#Check
df.info()
```

#### Fixing `postal-code`

```python
#To remove trialing .0 converted into int first
df['postal-code'] = df['postal-code'].astype('int')
```

```python
# Now comvert into str
df['postal-code'] = df['postal-code'].astype('str')
```

```python
df['postal-code'].sample()
```

```python
# For postal code that are of length 4 we have to make it 5 by filling leading 0
df['postal-code'] = df['postal-code'].str.zfill(5)
```

```python
df['postal-code'].sample()
```

#### Fixing `Ship Mode`, `Segment`, `Region`, `Category` & `Sub-Category`.

```python
df['ship-mode'] = df['ship-mode'].astype('category')
df['segment'] = df['segment'].astype('category')
df['region'] = df['region'].astype('category')
df['category'] = df['category'].astype('category')
df['sub-category'] = df['sub-category'].astype('category')
```

```python
# Check
df.info()
```

### Drop `row-id` column
It is not usefull column for analysis so droping it

```python
df.drop(columns='row-id', inplace=True)
```
