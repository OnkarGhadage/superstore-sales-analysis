```python
df.head()
```

---

**`superstore-sales`**

`Row ID` - It is just a number representing the row number. We have 9800 rows.  
`Order ID` - It is the order ID, It is the same for those product which are ordered at the same time by the same customer.  
`Order Date` - The date of ordered is in the Order Date column  
`Ship Date` - This column contains the date. The date is of the day on which the product is delivered to customer.  
`Ship Mode` - Ship mode column contains the string values. Which have represents the shiping mode of the product.  
`Customer ID` - Customer ID in alphanumeric code. This is the combination of name and surname initials with a unique code concatinated with '-'  
`Customer Name` - Name of the customer.  
`Segment` - This column contains the user segment. The customers are splited into 3 groups. (**Consumer**: Regular people buying things for themselves. **Corporate**: Big companies buying things for their offices. **Home Office**: People who work from home buying things for their workspace.)  
`Country` - Country name which is only United States.
`City` - It has the city name. Where the customer is live or package is delivered.  
`State` - The state.  
`Postal Code` - The postal code.  
`Region` - The US geographical area is splited into 4 parts (Central, East, South & West)  
`Product ID` - It is the unique id for each product. It is like (category-subcategoey-uniquecode)  
`Category` - It has the category of the product. (ex. Furniture, Offie Supplies, Art, ...etc)  
`Sub-Category` - It has the sub-category of the product. (ex. Bookcases, chair, Table, ...etc)  
`Product Name` - The actual name of the product by the brand.  
`Sales` - It is the price of the product.   
`Quantity` - It is the number of particular product that customer order.  
`Discount` - The discount given (e.g., 0.2 = 20%).  
`Profit` - It is the profit made by the superstore, negative value mean loss.
